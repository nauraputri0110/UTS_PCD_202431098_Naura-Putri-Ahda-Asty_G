```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

# MEMBACA GAMBAR

```python
img = cv2.imread('foto intan.jpeg')
```

# Konversi BGR ke RGB

```python
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

# GRAYSCALE

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

# GRAY DIPERCERAH

```python
bright = cv2.convertScaleAbs(gray, alpha=1.0, beta=50)
```

# GRAY DIPERKONTRAS

```python
contrast = cv2.convertScaleAbs(gray, alpha=2.0, beta=0)
```

# GRAY DIPERCERAH +
# DIPERKONTRAS

```python
bright_contrast = cv2.convertScaleAbs(gray, alpha=2.0, beta=50)
```

# MENAMPILKAN HASIL

```python
plt.figure(figsize=(15,10))

# Gambar Asli
plt.subplot(2,3,1)
plt.imshow(rgb)
plt.title("Gambar Asli")
plt.axis('off')

# Gambar Gray
plt.subplot(2,3,2)
plt.imshow(gray, cmap='gray')
plt.title("Gambar Gray")
plt.axis('off')

# Gray Dipercerah
plt.subplot(2,3,3)
plt.imshow(bright, cmap='gray')
plt.title("Gray Dipercerah")
plt.axis('off')

# Gray Diperkontras
plt.subplot(2,3,4)
plt.imshow(contrast, cmap='gray')
plt.title("Gray Diperkontras")
plt.axis('off')

# Gray Dipercerah + Diperkontras
plt.subplot(2,3,5)
plt.imshow(bright_contrast, cmap='gray')
plt.title("Gray Dipercerah & Diperkontras")
plt.axis('off')

plt.tight_layout()
plt.show()
```
